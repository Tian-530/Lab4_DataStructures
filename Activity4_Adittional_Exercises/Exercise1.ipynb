```python
# [CELL 1] STUDENT CONFIGURATION

# Student inputs
LAST_NAME = "Obiedo"
STUDENT_ID = "TUPM-26-4134"
FAVORITE_ARTIST = "Cup of Joe"

# Derived values
SEED_NUM = int(STUDENT_ID[-1])
artist_letters = [c for c in FAVORITE_ARTIST.upper() if c.isalpha()]
letter_values = [ord(c) - 64 for c in artist_letters]  # A=1 ... Z=26

print("Last Name:", LAST_NAME)
print("Student ID:", STUDENT_ID)
print("Favorite Artist:", FAVORITE_ARTIST)
print("SEED_NUM:", SEED_NUM)
print("Artist Letters:", artist_letters)
print("Letter Values:", letter_values)
```

```python
# [CELL 2]  TEST SOURCES AND CATEGORIES

from collections import defaultdict

# Test sources and categories
sources = [f"{LAST_NAME}-T{n}" for n in range(1, 6)]
categories = ("TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED")
```

```python
# [CELL 3]  GENERATED TEST DATA

# Generated test records: (source, category, value)
records = []
for i in range(10):
    value = letter_values[i % len(letter_values)] * SEED_NUM + i
    records.append((sources[i % 5], categories[(i // 2) % 5], value))

# Repeated record
records.append(records[0])

print("=== GENERATED TEST DATA ===")
for record in records:
    print(record)
```

```python
#[CELL 4]  ORGANIZED TEST DATA

# Organized test data (by source, then category)
organized = {source: defaultdict(list) for source in sources}
for source, category, value in records:
    organized[source][category].append(value)

print("=== ORGANIZED TEST DATA ===")
for source, category_values in organized.items():
    print(source, dict(category_values))
```

```python
# [CELL 5]  REPEATED/DISTINCT DATA

# Repeated and distinct data
record_count = defaultdict(int)
for record in records:
    record_count[record] += 1
repeated = [record for record, count in record_count.items() if count > 1]

unique_records = []
seen = set()
for record in records:
    if record not in seen:
        seen.add(record)
        unique_records.append(record)

distinct_sources = {record[0] for record in unique_records}
distinct_categories = {record[1] for record in unique_records}

print("=== REPEATED / DISTINCT DATA ===")
print("Repeated Records:", repeated)
print("Distinct Records:", len(unique_records))
print("Distinct Sources:", len(distinct_sources))
print("Distinct Categories:", len(distinct_categories))
```

```python
#[CELL 6]  NUMERICAL AND CATEGORICAL SUMMARIES

# Numerical and categorical summaries
values_by_category = defaultdict(list)
records_per_source = defaultdict(int)
records_per_category = defaultdict(int)
for source, category, value in unique_records:
    values_by_category[category].append(value)
    records_per_source[source] += 1
    records_per_category[category] += 1

summary = {}
for category, values in values_by_category.items():
    summary[category] = {
        "count": len(values),
        "min": min(values),
        "max": max(values),
        "total": sum(values),
        "average": round(sum(values) / len(values), 2),
    }

print("=== NUMERICAL AND CATEGORICAL SUMMARIES ===")
for category, info in summary.items():
    print(f"{category}: {info}")
print("Records per Source:", dict(records_per_source))
print("Records per Category:", dict(records_per_category))
```

```python
# [CELL 7]  PROCESSED RESULTS

# Processed results (compared with category average)
processed = []
for source, category, value in unique_records:
    if value > summary[category]["average"]:
        level = "ABOVE AVERAGE"
    else:
        level = "AT/BELOW AVERAGE"
    processed.append((source, category, value, level))

print("=== PROCESSED RESULTS ===")
for item in processed:
    print(item)
```

```python
# [CELL 8]  ORDERED RESULTS

# Ordered results (highest first)
ordered_records = [(value, source, category) for source, category, value, level in processed]
ordered_records.sort(reverse=True)
ordered_categories = [(info["average"], category) for category, info in summary.items()]
ordered_categories.sort(reverse=True)

print("=== ORDERED RESULTS ===")
print("Records by Value:", ordered_records)
print("Categories by Average:", ordered_categories)
```

```python
# [CELL 9]  FINAL ORGANIZED AND SUMMARIZED TEST DATASET

# Final organized and summarized test dataset
print("=== FINAL ORGANIZED AND SUMMARIZED TEST DATASET ===")
print(f"{'Source':<12}{'Category':<14}{'Value':<8}{'Level'}")
for source, category, value, level in processed:
    print(f"{source:<12}{category:<14}{value:<8}{level}")
print(f"\nOriginal Records: {len(records)}")
print(f"Processed Records: {len(processed)}")
print(f"Repeated Records: {len(repeated)}")
print(f"Highest Measurement: {ordered_records[0]}")
print(f"Lowest Measurement: {ordered_records[-1]}")
print(f"Highest Average Category: {ordered_categories[0][1]} ({ordered_categories[0][0]})")
print(f"Lowest Average Category: {ordered_categories[-1][1]} ({ordered_categories[-1][0]})")
```

```python
# [CELL 1] STUDENT CONFIGURATION

# Student inputs
LAST_NAME = "Obiedo"
STUDENT_ID = "TUPM-26-4134"
FAVORITE_ARTIST = "Cup of Joe"

# Derived values
SEED_NUM = int(STUDENT_ID[-1])
artist_letters = [c for c in FAVORITE_ARTIST.upper() if c.isalpha()]
letter_values = [ord(c) - 64 for c in artist_letters]  # A=1 ... Z=26

print("Last Name:", LAST_NAME)
print("Student ID:", STUDENT_ID)
print("Favorite Artist:", FAVORITE_ARTIST)
print("SEED_NUM:", SEED_NUM)
print("Artist Letters:", artist_letters)
print("Letter Values:", letter_values)
```

```python
# [CELL 2]  CUSTOMERS, PRODUCTS, AND PRICES

from array import array
from collections import defaultdict, deque

# Customers, products, and fixed prices
customers = [f"{LAST_NAME}-C{n}" for n in range(1, 6)]
products = ("NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR")
prices = {"NOTEBOOK": 45, "PEN": 10, "USB": 250, "HEADPHONES": 800, "CALCULATOR": 600}
```

```python
# [CELL 3]  GENERATED TRANSACTION RECORDS

# Quantities from SEED_NUM and artist letters
quantities = array('i', [(letter_values[t % len(letter_values)] * SEED_NUM) % 5 + 1 for t in range(8)])

# Generated transaction records
transactions = []
for t in range(8):
    transactions.append({
        "customer": customers[t % 5],
        "product": products[(t + SEED_NUM) % 5],
        "quantity": quantities[t],
    })

print("=== GENERATED TRANSACTION RECORDS ===")
for transaction in transactions:
    print(transaction)
```

```python
# [CELL 4]  ORGANIZED TRANSACTION DATA

# Organized transaction data with values
queue = deque(transactions)
organized = []
while queue:
    transaction = queue.popleft()
    transaction["value"] = transaction["quantity"] * prices[transaction["product"]]
    organized.append(transaction)

print("=== ORGANIZED TRANSACTION DATA ===")
for transaction in organized:
    print(f"{transaction['customer']} | {transaction['product']} | "
          f"qty={transaction['quantity']} | value={transaction['value']}")
```

```python
# [CELL 5]  REPEATED/DISTINCT TRANSACTIONS

# Repeated and distinct transactions
pair_count = defaultdict(int)
for transaction in organized:
    pair_count[(transaction["customer"], transaction["product"])] += 1

repeated = [pair for pair, count in pair_count.items() if count > 1]
distinct_customers = {t["customer"] for t in organized}
distinct_products = {t["product"] for t in organized}

print("=== REPEATED / DISTINCT TRANSACTIONS ===")
print("Repeated Customer-Product Pairs:", repeated)
print("Distinct Customers:", len(distinct_customers))
print("Distinct Products:", len(distinct_products))
```

```python
# [CELL 6]  TRANSACTION VALUES

# Transaction values
values = [t["value"] for t in organized]
overall_total = sum(values)

print("=== TRANSACTION VALUES ===")
print("Values:", values)
print("Overall Total:", overall_total)
```

```python
# [CELL 7]  CUSTOMER AND PRODUCT SUMMARIES

# Customer and product summaries
customer_total = defaultdict(int)
customer_count = defaultdict(int)
product_units = defaultdict(int)
product_total = defaultdict(int)
for t in organized:
    customer_total[t["customer"]] += t["value"]
    customer_count[t["customer"]] += 1
    product_units[t["product"]] += t["quantity"]
    product_total[t["product"]] += t["value"]

print("=== CUSTOMER SUMMARIES ===")
for customer in customer_total:
    print(f"{customer}: transactions={customer_count[customer]}, total={customer_total[customer]}")

print("\n=== PRODUCT SUMMARIES ===")
for product in product_total:
    print(f"{product}: units={product_units[product]}, total={product_total[product]}")
```

```python
# [CELL 8]  ORDERED RESULTS

# Ordered results (highest total first)
customer_ranking = [(total, customer) for customer, total in customer_total.items()]
customer_ranking.sort(reverse=True)
product_ranking = [(total, product) for product, total in product_total.items()]
product_ranking.sort(reverse=True)

print("=== ORDERED RESULTS ===")
print("Customers:", customer_ranking)
print("Products:", product_ranking)
```

```python
# [CELL 9]  OVERALL TRANSACTION SUMMARY

# Overall transaction summary
largest = organized[values.index(max(values))]
print("=== OVERALL TRANSACTION SUMMARY ===")
print("Total Transactions:", len(organized))
print("Total Value:", overall_total)
print("Average Value:", round(overall_total / len(organized), 2))
print("Largest Transaction:", largest["customer"], largest["product"], largest["value"])
```

```python
# [CELL 10]  FINAL CONSOLIDATED TRANSACTION REPORT

# Final Consolidated Transaction Report
print("=== FINAL CONSOLIDATED TRANSACTION REPORT ===")
print(f"{'Rank':<6}{'Customer':<14}{'Total Value'}")
for rank, (total, customer) in enumerate(customer_ranking, 1):
    print(f"{rank:<6}{customer:<14}{total}")
print(f"\nTop Product: {product_ranking[0][1]} ({product_ranking[0][0]})")
print(f"Repeated Purchases: {len(repeated)}")
print(f"Overall Total: {overall_total}")
```

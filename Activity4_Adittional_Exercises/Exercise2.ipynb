```python
# [CELL 1] STUDENT CONFIGURATION

# Student inputs
LAST_NAME = "Obiedo"
STUDENT_ID = "TUPM-26-4134"
FAVORITE_ARTIST = "Cup of Joe"

# Derived values
SEED_NUM = int(STUDENT_ID[-1])
artist_letters = [c for c in FAVORITE_ARTIST.upper() if c.isalpha()]
letter_values = [ord(c) - 64 for c in artist_letters]  # A=1 ... Z=26

print("Last Name:", LAST_NAME)
print("Student ID:", STUDENT_ID)
print("Favorite Artist:", FAVORITE_ARTIST)
print("SEED_NUM:", SEED_NUM)
print("Artist Letters:", artist_letters)
print("Letter Values:", letter_values)
```

```python
# [CELL 2]  STUDENTS AND SUBJECTS

from collections import defaultdict

# Students and subjects
students = [f"{LAST_NAME}-S{n}" for n in range(1, 6)]
subjects = ("PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH")
```

```python
# [CELL 3]  GENERATED PERFORMANCE RECORDS

# Generated performance records: (student, subject, score)
records = []
position = 0
for student in students:
    for subject in subjects:
        value = letter_values[position % len(letter_values)]
        score = 55 + (value * SEED_NUM + position) % 46
        records.append((student, subject, score))
        position += 1

# Repeated record
records.append(records[0])

# Incomplete record (missing score)
records[13] = (records[13][0], records[13][1], None)

print("=== GENERATED PERFORMANCE RECORDS ===")
for record in records:
    print(record)
```

```python
# [CELL 4]  ORGANIZED STUDENT/SUBJECT DATA

# Organized student/subject data
organized = defaultdict(dict)
pair_count = defaultdict(int)
for student, subject, score in records:
    pair_count[(student, subject)] += 1
    organized[student][subject] = score

print("=== ORGANIZED STUDENT/SUBJECT DATA ===")
for student, subject_scores in organized.items():
    print(student, subject_scores)
```

```python
# [CELL 5]  REPEATED/INCOMPLETE RECORDS

# Repeated and incomplete records
repeated = [pair for pair, count in pair_count.items() if count > 1]
incomplete = [(student, subject) for student, subject, score in records if score is None]

print("=== REPEATED / INCOMPLETE RECORDS ===")
print("Repeated:", repeated)
print("Incomplete:", incomplete)
```

```python
# [CELL 6]  PERFORMANCE CATEGORIES

# Performance categories
def get_category(score):
    if score >= 90:
        return "EXCELLENT"
    elif score >= 80:
        return "GOOD"
    elif score >= 70:
        return "SATISFACTORY"
    else:
        return "NEEDS IMPROVEMENT"

categories = {}
for student, subject_scores in organized.items():
    for subject, score in subject_scores.items():
        if score is not None:
            categories[(student, subject)] = get_category(score)

print("=== PERFORMANCE CATEGORIES ===")
for (student, subject), category in categories.items():
    print(f"{student} | {subject}: {category}")
```

```python
# [CELL 7]  STUDENT AND SUBJECT SUMMARIES

# Student and subject summaries
student_scores = defaultdict(list)
subject_scores_all = defaultdict(list)
for student, subject_scores in organized.items():
    for subject, score in subject_scores.items():
        if score is not None:
            student_scores[student].append(score)
            subject_scores_all[subject].append(score)

student_avg = {student: sum(s) / len(s) for student, s in student_scores.items()}
subject_avg = {subject: sum(s) / len(s) for subject, s in subject_scores_all.items()}

print("=== STUDENT AND SUBJECT SUMMARIES ===")
for student, avg in student_avg.items():
    print(f"{student}: average={avg:.2f}, valid scores={len(student_scores[student])}")
for subject, avg in subject_avg.items():
    print(f"{subject}: average={avg:.2f}, highest={max(subject_scores_all[subject])}")
```

```python
# [CELL 8]  ORDERED RESULTS

# Ordered results (highest average first)
student_ranking = [(round(avg, 2), student) for student, avg in student_avg.items()]
student_ranking.sort(reverse=True)
subject_ranking = [(round(avg, 2), subject) for subject, avg in subject_avg.items()]
subject_ranking.sort(reverse=True)

print("=== ORDERED RESULTS ===")
print("Students:", student_ranking)
print("Subjects:", subject_ranking)
```

```python
# [CELL 9]  FINAL STUDENT PERFORMANCE REPORT

# Final Student Performance Report
print("=== FINAL STUDENT PERFORMANCE REPORT ===")
print(f"{'Rank':<6}{'Student':<14}{'Average':<10}{'Category'}")
for rank, (avg, student) in enumerate(student_ranking, 1):
    print(f"{rank:<6}{student:<14}{avg:<10}{get_category(avg)}")
print(f"\nTotal Records: {len(records)}")
print(f"Valid Scores: {len(categories)}")
print(f"Repeated Records: {len(repeated)}")
print(f"Incomplete Records: {len(incomplete)}")
print(f"Top Student: {student_ranking[0][1]}")
print(f"Strongest Subject: {subject_ranking[0][1]}")
print(f"Weakest Subject: {subject_ranking[-1][1]}")
```
